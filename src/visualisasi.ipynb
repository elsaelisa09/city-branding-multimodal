```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Nilai probabilitas dari hasil analisis
labels = ['Top 1', 'Top 2', 'Top 3']
probabilities = [0.737916756, 0.181100511, 0.063329542]
colors = ['#2563EB', '#14B8A6', '#F59E0B']

fig, ax = plt.subplots(figsize=(9, 5.5))
bars = ax.bar(labels, probabilities, color=colors, width=0.6)

# Tampilkan nilai desimal dan persentase di atas setiap bar
for bar, probability in zip(bars, probabilities):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.02,
        f'{probability:.3f}\n({probability:.2%})',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

ax.set_title('Rata-rata Probability', fontsize=18, fontweight='bold', pad=18)
ax.set_xlabel('Peringkat Prediksi', fontsize=12)
ax.set_ylabel('Probability', fontsize=12)
ax.set_ylim(0, 0.85)
ax.grid(axis='y', linestyle='--', alpha=0.35)
ax.set_axisbelow(True)

# Legend menjelaskan warna dan nilai tiap kategori
legend_handles = [
    Patch(facecolor=color, label=f'{label}: {probability:.9f}')
    for label, probability, color in zip(labels, probabilities, colors)
]
ax.legend(
    handles=legend_handles,
    title='Kategori dan Nilai Probability',
    loc='upper right',
    frameon=True
)

plt.tight_layout()
plt.show()
```
