```python
import pandas as pd
import os

# ==============================
# 1. Load dataset
# ==============================

df = pd.read_csv(
    "../Scraping Data Zefanya (10 Aug )/newdataset-cleaning.csv"
)

print("Jumlah baris dataset:", len(df))
print("Jumlah FileName terisi:", df['FileName'].notna().sum())
print("Jumlah FileName kosong:", df['FileName'].isna().sum())


# ==============================
# 2. Load folder instagram_images
# ==============================

image_folder = "../Scraping Data Zefanya (10 Aug )/Image"

print("\nApakah folder ditemukan:", os.path.exists(image_folder))
print("Apakah path merupakan folder:", os.path.isdir(image_folder))


# ==============================
# 3. Baca semua file
# ==============================

all_files = os.listdir(image_folder)

print("\nTotal isi folder:", len(all_files))


# ==============================
# 4. Hanya hitung file gambar
# ==============================

image_extensions = (".jpg", ".jpeg", ".png", ".webp")

image_files = [
    file for file in all_files
    if file.lower().endswith(image_extensions)
]

print("Total file gambar:", len(image_files))


# ==============================
# 5. Tampilkan 10 contoh file
# ==============================

print("\nContoh file gambar:")

for file in image_files[:10]:
    print(file)
```

```python
import pandas as pd
import os
import shutil

# ==========================================
# 1. Load dataset
# ==========================================

df = pd.read_csv(
    "../Scraping Data Zefanya (10 Aug )/newdataset-cleaning.csv"
)

# Ambil FileName yang tidak kosong
dataset_filenames = set(
    df['FileName']
    .dropna()
    .astype(str)
    .str.strip()
)

print("Jumlah FileName di dataset:", len(dataset_filenames))


# ==========================================
# 2. Folder sumber gambar
# ==========================================

image_folder = "../Scraping Data Zefanya (10 Aug )/Image"


# ==========================================
# 3. Buat folder baru
# ==========================================

output_folder = "../Scraping Data Zefanya (10 Aug )/New Images Clean"

os.makedirs(output_folder, exist_ok=True)

print("Folder output:", output_folder)


# ==========================================
# 4. Ambil file dari folder instagram_images
# ==========================================

image_files = [
    file
    for file in os.listdir(image_folder)
    if os.path.isfile(os.path.join(image_folder, file))
]

print("Jumlah file di folder instagram_images:", len(image_files))


# ==========================================
# 5. Cari file yang cocok
# ==========================================

matched_files = []

for file in image_files:

    if file in dataset_filenames:
        matched_files.append(file)

        source_path = os.path.join(image_folder, file)
        destination_path = os.path.join(output_folder, file)

        shutil.copy2(source_path, destination_path)


# ==========================================
# 6. Hasil
# ==========================================

print("\n=== HASIL MATCHING ===")
print("File cocok:", len(matched_files))
print("File tidak cocok:", len(image_files) - len(matched_files))
print("File di folder New Images Clean:", len(os.listdir(output_folder)))
```

```python
# Cari FileName dari dataset yang tidak ditemukan di folder instagram_images

dataset_filenames = set(
    df['FileName']
    .dropna()
    .astype(str)
    .str.strip()
)

image_filenames = set(image_files)

missing_from_folder = dataset_filenames - image_filenames

print("Jumlah FileName dataset yang tidak ditemukan:", len(missing_from_folder))

print("\nFileName yang tidak ditemukan:")
for file in missing_from_folder:
    print(file)
```

```python
import pandas as pd
import os

# ==========================================
# 1. Load CSV final
# ==========================================

df_final = pd.read_csv(
    "../Scraping Data Zefanya (10 Aug )/newdataset-cleaning-final.csv"
)

# Ambil FileName dari CSV
csv_filenames = set(
    df_final['FileName']
    .dropna()
    .astype(str)
    .str.strip()
)


# ==========================================
# 2. Load folder New Images Clean
# ==========================================

image_folder = "../Scraping Data Zefanya (10 Aug )/New Images Clean"

image_files = {
    file
    for file in os.listdir(image_folder)
    if os.path.isfile(os.path.join(image_folder, file))
}


# ==========================================
# 3. Cek jumlah
# ==========================================

print("Jumlah FileName di CSV       :", len(csv_filenames))
print("Jumlah file di folder gambar :", len(image_files))


# ==========================================
# 4. Cek file yang ada di CSV tapi
#    tidak ada di folder
# ==========================================

missing_images = csv_filenames - image_files

print("\nCSV ada, tetapi gambar tidak ada :", len(missing_images))

if missing_images:
    print("\nFile yang tidak ditemukan:")
    for file in sorted(missing_images):
        print(file)


# ==========================================
# 5. Cek file yang ada di folder tetapi
#    tidak ada di CSV
# ==========================================

extra_images = image_files - csv_filenames

print("\nGambar ada, tetapi tidak ada di CSV :", len(extra_images))

if extra_images:
    print("\nFile yang tidak ada di CSV:")
    for file in sorted(extra_images):
        print(file)


# ==========================================
# 6. Kesimpulan
# ==========================================

if csv_filenames == image_files:
    print("\n✅ MATCH 100%")
    print("Semua FileName di CSV memiliki gambar yang sesuai.")
else:
    print("\n❌ BELUM MATCH 100%")
```
