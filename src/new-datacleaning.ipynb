```python
import pandas as pd
from pathlib import Path

# Baca CSV
df = pd.read_csv("newdataset-cleaning-final.csv")

# Folder gambar
folder_instagram = Path("instagram_images")

# Ambil semua nama file di folder instagram_images
file_folder = {
    f.name
    for f in folder_instagram.iterdir()
    if f.is_file()
}

# Ambil semua nama file dari kolom FileName di CSV
file_csv = set(
    df["FileName"]
    .dropna()
    .astype(str)
    .str.strip()
)

# Cari file yang ada di folder tetapi TIDAK ada di CSV
file_tidak_ada_di_csv = sorted(file_folder - file_csv)

print("Jumlah file di folder instagram_images :", len(file_folder))
print("Jumlah FileName unik di CSV             :", len(file_csv))
print("Jumlah file folder yang tidak ada di CSV :", len(file_tidak_ada_di_csv))

if len(file_tidak_ada_di_csv) == 0:
    print("\n✅ Semua file di folder instagram_images ada di kolom FileName CSV.")
else:
    print("\n❌ Ada file di folder instagram_images yang tidak ditemukan di CSV.")
```

```python
import pandas as pd
from pathlib import Path

# Baca dataset utama
df = pd.read_csv("newdataset-cleaning-final.csv")

# Tentukan folder gambar
folder_images = Path("instagram_images")

# Ambil semua nama file yang ada di folder instagram_images
nama_file_folder = {
    file.name.strip()
    for file in folder_images.iterdir()
    if file.is_file()
}

# Bersihkan kolom FileName pada dataset
df["FileName"] = df["FileName"].astype(str).str.strip()

# Filter baris berdasarkan FileName yang ada di folder instagram_images
df_instagram = df[
    df["FileName"].isin(nama_file_folder)
].copy()

# Simpan hasil ke CSV baru
output_file = "instagram_images.csv"
df_instagram.to_csv(output_file, index=False)

# Tampilkan hasil pengecekan
print("===== HASIL =====")
print("Jumlah file di instagram_images   :", len(nama_file_folder))
print("Jumlah baris dataset awal         :", len(df))
print("Jumlah baris instagram_images.csv :", len(df_instagram))
print("Jumlah kolom                      :", len(df_instagram.columns))

# Cek apakah semua file di folder ditemukan di dataset
file_csv_hasil = set(df_instagram["FileName"])
file_tidak_ditemukan = nama_file_folder - file_csv_hasil

if len(file_tidak_ditemukan) == 0:
    print("\nSemua file di instagram_images ditemukan di dataset.")
else:
    print(
        f"\nAda {len(file_tidak_ditemukan)} file "
        "di instagram_images yang tidak ditemukan di dataset."
    )

print("\nFile tersimpan sebagai:")
print(output_file)
```
