```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
import requests
import os
from pathlib import Path
from tqdm import tqdm

csv_path = "newdataset-cleaning.csv"

df = pd.read_csv(
    csv_path,
    dtype={
        "id": "string",
        "owner/id": "string",
        "location/id": "string"
    }
)

print("Jumlah data:", len(df))

image_dir = Path("newImages")
image_dir.mkdir(exist_ok=True)

file_names = []
download_status = []

for i, image_url in tqdm(
    enumerate(df["image/url"]),
    total=len(df),
    desc="Downloading images"
):

    # Nama file - nomor baris
    file_name = f"image_{i+1:06d}.jpg"
    file_path = image_dir / file_name

    try:
        response = requests.get(
            image_url,
            timeout=30
        )

        response.raise_for_status()

        with open(file_path, "wb") as f:
            f.write(response.content)

        file_names.append(file_name)
        download_status.append("success")

    except Exception as e:
        print(f"\nGagal download baris {i}: {e}")

        file_names.append(file_name)
        download_status.append("failed")

df["FileName"] = file_names
df["DownloadStatus"] = download_status

output_csv = "dataset_with_images.csv"

df.to_csv(
    output_csv,
    index=False
)

print("\nSelesai!")
print("CSV :", output_csv)
print("Folder gambar :", image_dir)
print("Berhasil :", download_status.count("success"))
print("Gagal    :", download_status.count("failed"))
```
